```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headset"]
sales = [50, 120, 80, 40, 100]
```

```python
# Line Chart

plt.figure()
plt.plot(products, sales, marker="o") 

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Units Sold")

# Data callout
plt.annotate(
    "Highest Sales",
    xy=("Mouse", 120),
    xytext=("Laptop", 140),
    arrowprops=dict(arrowstyle="->")
)

plt.savefig("line_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Bar Chart

plt.figure()
plt.bar(products, sales)   

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Units Sold")

plt.savefig("bar_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Scatter Plot

price = [800, 20, 50, 300, 60]
plt.figure()
plt.scatter(price, sales)

plt.title("Price vs Sales")
plt.xlabel("Price")
plt.ylabel("Units Sold")

plt.savefig("scatter_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Histogram

plt.figure()
plt.hist(sales, bins=5)

plt.title("Distribution of Sales")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")

plt.savefig("histogram.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Box Plot

plt.figure()
plt.boxplot(sales)

plt.title("Sales Distribution")
plt.ylabel("Units Sold")

plt.savefig("boxplot.png", dpi=300, bbox_inches="tight")
plt.show()
```
