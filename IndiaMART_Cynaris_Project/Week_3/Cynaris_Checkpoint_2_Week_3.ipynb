```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("da_sales_transactions.csv")
customers = pd.read_csv("da_customer_data.csv")
products = pd.read_csv("da_product_catalogue.csv")
```

```python
print("Sales:", sales.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
```

```python
print("Sales missing values:")
print(sales.isnull().sum())

print("\nCustomer missing values:")
print(customers.isnull().sum())

print("\nProduct missing values:")
print(products.isnull().sum())
```

```python
print("Sales duplicates:", sales.duplicated().sum())
print("Customer duplicates:", customers.duplicated().sum())
print("Product duplicates:", products.duplicated().sum())
```

```python
print("===== SALES DATA =====")
sales.info()
print("\nSales Statistical Summary:")
display(sales.describe())

print("\n===== CUSTOMER DATA =====")
customers.info()
print("\nCustomer Statistical Summary:")
display(customers.describe())

print("\n===== PRODUCT DATA =====")
products.info()
print("\nProduct Statistical Summary:")
display(products.describe())
```

```python
sales["Gross_Sales"] = sales["Units"] * sales["Unit_Price"]

sales["Discount_Amount"] = (
    sales["Gross_Sales"] * sales["Discount_Pct"] / 100
)

sales["Net_Sales"] = (
    sales["Gross_Sales"] - sales["Discount_Amount"]
)

sales.head()
```

```python
print("Total Transactions:", len(sales))
print("Total Units Sold:", sales["Units"].sum())
print("Gross Sales:", sales["Gross_Sales"].sum())
print("Net Sales:", sales["Net_Sales"].sum())
print("Average Transaction Value:", sales["Net_Sales"].mean())
print("Average Discount:", sales["Discount_Pct"].mean())
print("Return Rate:", sales["Return_Flag"].mean() * 100)
```

```python
churn_rate = customers["Churned"].mean() * 100

print("Customer Churn Rate:", churn_rate)
```

```python
channel_sales = sales.groupby("Channel").agg(
    Transactions=("Transaction_ID", "count"),
    Net_Sales=("Net_Sales", "sum"),
    Return_Rate=("Return_Flag", "mean")
).reset_index()

channel_sales["Return_Rate"] = channel_sales["Return_Rate"] * 100

channel_sales
```

```python
category_sales = sales.groupby("Category").agg(
    Transactions=("Transaction_ID", "count"),
    Units=("Units", "sum"),
    Net_Sales=("Net_Sales", "sum"),
    Return_Rate=("Return_Flag", "mean")
).reset_index()

category_sales["Return_Rate"] = category_sales["Return_Rate"] * 100

category_sales.sort_values("Net_Sales", ascending=False)
```

```python
analysis_df = sales.merge(
    customers,
    on="Customer_ID",
    how="left"
)

analysis_df = analysis_df.merge(
    products,
    on="Product_ID",
    how="left"
)

analysis_df.head()
```

```python
print("Combined dataset shape:", analysis_df.shape)
```

```python
print(
    "Customer match rate:",
    sales["Customer_ID"].isin(customers["Customer_ID"]).mean() * 100
)

print(
    "Product match rate:",
    sales["Product_ID"].isin(products["Product_ID"]).mean() * 100
)
```

```python
analysis_df.to_csv(
    "cleaned_sales_customer_product.csv",
    index=False
)

print("Cleaned dataset exported successfully.")
```
